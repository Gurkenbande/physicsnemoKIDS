```python
import matplotlib.pyplot as plt

pred = ds["prediction"]["output"]

# Beispiel: sample 0, time 0
field = pred.isel(sample=0, time=0)

plt.figure()
plt.imshow(field, origin="lower")
plt.colorbar(label="Value")
plt.title("Prediction (sample 0, time 0)")
plt.show()
```

```python
pred = ds["prediction"]["output"].isel(sample=0, time=0)
truth = ds["truth"]["output"].isel(time=0)

fig, axs = plt.subplots(1, 3, figsize=(12, 4))

pred.plot(ax=axs[0], cmap="viridis")
axs[0].set_title("Prediction")

truth.plot(ax=axs[1], cmap="viridis")
axs[1].set_title("Ground Truth")

(pred - truth).plot(ax=axs[2], cmap="coolwarm")
axs[2].set_title("Error (Pred - Truth)")

plt.tight_layout()
plt.show()
```

```python
import xarray as xr
import matplotlib.pyplot as plt

ds = xr.open_dataset("output/test_results.nc")

pred = ds["prediction"]["output"]  # dims: sample, time, lat, lon
truth = ds["truth"]["output"]

sample_idx = 0
time_idx = 0

fig, axs = plt.subplots(1, 3, figsize=(12, 4))

def update():
    axs[0].clear()
    axs[1].clear()
    axs[2].clear()

    p = pred.isel(sample=sample_idx, time=time_idx)
    t = truth.isel(time=time_idx)

    p.plot(ax=axs[0])
    axs[0].set_title(f"Prediction | sample={sample_idx}, time={time_idx}")

    t.plot(ax=axs[1])
    axs[1].set_title("Ground Truth")

    (p - t).plot(ax=axs[2])
    axs[2].set_title("Error")

    plt.tight_layout()
    fig.canvas.draw_idle()

def on_key(event):
    global time_idx

    if event.key == "right":
        time_idx = (time_idx + 1) % pred.sizes["time"]
        update()
    elif event.key == "left":
        time_idx = (time_idx - 1) % pred.sizes["time"]
        update()

fig.canvas.mpl_connect("key_press_event", on_key)

update()
plt.show()
```
